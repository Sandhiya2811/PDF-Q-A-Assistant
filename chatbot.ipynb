```python
from langchain_community.document_loaders import PyMuPDFLoader

file_path = "Interview Preparation Material on AI ML DL  NLP.pdf"
loader = PyMuPDFLoader(file_path)
docs = loader.load()
```

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter

text_splitter = RecursiveCharacterTextSplitter(chunk_size=1000, chunk_overlap=200)
docs_chunks = text_splitter.split_documents(docs)
docs_chunks
```

```python
from langchain_huggingface import HuggingFaceEmbeddings
from langchain_community.vectorstores import Chroma

# 1. Embeddings
embedding = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)

# 2. Store the text chunks in a vector store using ChromaDB
vectorstore = Chroma.from_documents(
    documents=docs_chunks,
    embedding=embedding,
    collection_name='MDT51_Vector_Collection',
    persist_directory="./chroma_db"  # Local-la save aagum
)

# 3. set the retriever for fast retrieval
retriever = vectorstore.as_retriever(search_kwargs={"k": 3})
```

```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.runnables import RunnablePassthrough


# =========================================================
# 1. Prompt
# =========================================================

prompt_template = """
You are an assistant for question-answering tasks.

Use the following pieces of retrieved context to answer the question.

If the answer is not present in the context, strictly respond with
"Information not found".

Do not use outside knowledge or assume anything.

Context:

{context}

Question: {question}

Answer:
"""

prompt = ChatPromptTemplate.from_template(prompt_template)


# =========================================================
# 2. RAG Chain
# =========================================================

rag_chain = (
    {
        "context": retriever,
        "question": RunnablePassthrough()
    }
    | prompt
    | llm
    | StrOutputParser()
)


# =========================================================
# 3. Get Question from User
# =========================================================

question = input("\nEnter your question: ").strip()


# =========================================================
# 4. Check Empty Question
# =========================================================

if not question:

    print("Please enter a question.")

else:

    # =====================================================
    # 5. Generate Answer
    # =====================================================

    response = rag_chain.invoke(question)


    # =====================================================
    # 6. Get Retrieved Documents + Scores
    # =====================================================

    results_with_scores = vectorstore.similarity_search_with_score(
        question,
        k=1
    )


    # =====================================================
    # 7. Display Answer
    # =====================================================

    print("\n" + "=" * 60)
    print("Answer:")
    print("=" * 60)

    print(response)


    # =====================================================
    # 8. Display Sources
    # =====================================================

    print("\n📄 Sources")
    print("-" * 60)

    for doc, score in results_with_scores:

        # ---------------------------------------------
        # Get page number
        # ---------------------------------------------

        page = doc.metadata.get("page", "Unknown")

        # ---------------------------------------------
        # Chroma score
        # ---------------------------------------------

        print(f"Page {page} | Distance: {score:.4f}")

    print("=" * 60)
```
